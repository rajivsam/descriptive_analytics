```python
import pandas as pd
fp = "../data/olist_prepared/SP_geo_revenue_by_city_2017.parquet"
df = pd.read_parquet(fp)
```

```python
df
```

```python
cities = df.index.tolist()
products = df.columns.tolist()
```

```python
from sklearn.decomposition import PCA
```

```python
NCOMP = 30
import numpy as np
ncomp = [i+1 for i in range(NCOMP)]
```

```python
ve = {}
for nc in ncomp:
    transformer = PCA(n_components=nc, random_state=0)
    transformer.fit(df)
    var_exp = np.sum(transformer.explained_variance_ratio_).item()
    var_exp = round(var_exp,3)
    ve[int(nc)] = var_exp
```

```python
import plotly.express as px
```

```python
dfve = pd.DataFrame.from_dict(ve, orient="index").reset_index()
```

```python
dfve.columns = ["nc", "var_exp"]
```

```python
# Create a line plot
fig = px.line(dfve, x='nc', y='var_exp', title='Variance Explained vs Number of Components', markers=True)
fig.show()
```
