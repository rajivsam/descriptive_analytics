```python
import pandas as pd
fp = "../data/sba_loans_data_raw/7a_504_foia_data_dictionary-asof-250331.xlsx"
df_dd = pd.read_excel(fp)
```

```python
sel_att = (df_dd["Field Name"] == "BorrName")
df_dd[sel_att]["Definition"]
```

```python
borrower = [ ("BorrName", "category"),  ("BorrCity", "category"), ("BorrState", "category"), ("BorrZip", "category"),\
             ("BusinessType", "category"), ("BusinessAge", "category")]
lender = [("BankName", "category"), ("BankCity", "category"), ("BankState", "category"), ("BankZip", "category")]
loan = [("InitialInterestRate", "float"), ("FixedOrVariableInterestInd", "category"), ("NaicsCode", "category"),\
        ("TermInMonths", "float"), ("RevolverStatus", "category"), ("CollateralInd", "category"),\
        ("SoldSecMrktInd", "category"), ("SoldSecMrktInd", "category"), ("GrossChargeOffAmount", "float"),\
        ("PaidInFullDate", "category"), ("ProjectCounty", "category"), ("ProjectState", "category"), ("LoanStatus", "category")]

comp_att_list = borrower + lender + loan
        
```

```python
 dtype_def = {a[0]: a[1] for a in comp_att_list}
att_list = list(dtype_def.keys())
```

```python
fp = "../data/sba_loans_data_raw/foia-7a-fy2020-present-asof-250331.tar.gz"
df = pd.read_csv(fp)
```

```python
df_req = df[att_list]
```

```python
df_req = df_req.astype(dtype_def)
```

```python
df_req.dtypes
```

```python
# get the definitions
entity_info = []
for a,t in dtype_def.items():
    att_meta_data = {}
    sel_att_defn = df_dd["Field Name"] == a
   
    att_meta_data["name"] = a
    att_meta_data["type"] = t
    att_meta_data["description"] =  df_dd[sel_att_defn]["Definition"].values[0]
    entity_info.append(att_meta_data)
```

```python
config_data = {}
config_data["raw_data"] = {}
config_data["raw_data"]["location"] = "../data/sba_loans_data_raw/foia-7a-fy2020-present-asof-250331.tar.gz"
config_data["raw_data"]["description"] = "Raw SBA loans data file"
config_data["raw_data"]["format"] = "CSV"
config_data["raw_data"]["encoding"] = "utf-8"
config_data["entity"] = entity_info
```

```python
import yaml
fp = "../config/sba_homog_explicit.yml"
with open(fp, 'w') as file:
    yaml.dump(config_data, file)
```
