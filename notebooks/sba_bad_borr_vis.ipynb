```python
import pandas as pd
fp = "../data/sba_loans_prepared/sba_loans_bad_borrowers_nbrh.csv"
df = pd.read_csv(fp)
```

```python
preds = df.columns.tolist()
preds = preds.remove("LoanStatus")
```

```python
import plotly.express as px
```

```python
df.index = "borr-" + df.index.astype(str)
```

```python
fig = px.imshow(df,
                labels=dict(x="NCA component", y="Borrower", color="Loan Status"),
                x= preds,
                y= df.index.tolist()
               )
fig.update_xaxes(side="top")
fig.show()
```
