```python
import pandas as pd
fp = "../data/olist_prepared/SP_geo_revenue_by_city_2017.parquet"
df = pd.read_parquet(fp)
```

```python
df
```

```python
import numpy as np
```

```python
def compute_row_pct(x):
    nzv = np.count_nonzero(x.values)
    pct = (nzv/df.shape[1])*100
    return pct
df["sparsity"]= df.apply(compute_row_pct , axis=1)
```

```python
df["sparsity"].hist()
```

```python
is_dense = df.sparsity > 10
df[is_dense]
```

```python
df = df[~is_dense]
```

```python
df
```

```python
cities = df.index.tolist()
products = df.columns.tolist()
if "sparisty" in products:
    products.remove("sparsity")
```

```python
df.loc[:,"revenue"] = df.apply(lambda x: sum(x[products].values), axis=1)
```

```python
qtls = np.linspace(.1, .9,9).tolist() + [.95]
```

```python
qtls
```

```python
df["revenue"].quantile(qtls)
```

```python
high_revenue_outliers = df.revenue > 7152
```

```python
df = df[~high_revenue_outliers]
```

```python
df["revenue"].hist()
```

```python
df[products].cov()
```

```python
from scipy.spatial.distance import pdist, squareform
```

```python
coords = df[products].values
distances = pdist(coords, metric='euclidean')

# Convert the condensed distance matrix to a squareform matrix
distance_matrix = squareform(distances)
```

```python
# Convert to Pandas DataFrame for better readability and indexing
distance_df = pd.DataFrame(distance_matrix, index=df.index, columns=df.index)
```
