```python
import pandas as pd
fp =  "../data/sba_loans_prepared/sba_loans_num_enc_train.csv"
df_train = pd.read_csv(fp)
```

```python
preds = df_train.columns.tolist()
preds.remove("LoanStatus")
contamination_train = df_train.LoanStatus.value_counts()[1]/df_train.LoanStatus.value_counts()[0]
```

```python
contamination_train
```

```python
select_chgoff = (df_train.LoanStatus == 1)
```

```python
df_train_good = df_train[~select_chgoff]
```

```python
df_train_chgoff = df_train[select_chgoff]
```

```python
X_df_good  = df_train_good[preds]
```

```python
X_df_bad = df_train_chgoff[preds]
```

```python
from sklearn.decomposition import PCA

import matplotlib.pyplot as plt


# Create a PCA instance with 2 components
pca = PCA(n_components=15)

# Fit PCA to the data and transform it
X_reduced = pca.fit_transform(X_df_good)
fp = "../data/sba_loans_prepared/sba_train_pca_good_loans.csv"
df_good_loans_pca = pd.DataFrame(X_reduced)
df_good_loans_pca.columns = ["PC-" + str(i+1) for i in range(15)]
df_good_loans_pca.to_csv(fp, index=False)



# Access explained variance ratio
print(f"Explained variance ratio: {pca.explained_variance_ratio_}")
print(f"Cumulative explained variance ratio: {pca.explained_variance_ratio_.sum()}")
```

```python
import numpy as np
df_good_loans_res = pd.DataFrame(pca.explained_variance_ratio_, columns =["exp_var"])
df_good_loans_res["cum_var_exp"] = df_good_loans_res["exp_var"].cumsum()
```

```python
df_good_loans_res
```

```python
fp_test = "../data/sba_loans_prepared/sba_loans_num_enc_test.csv"
df_test = pd.read_csv(fp_test)
X_df_test  = df_test[preds]
X_test_pca_good = pca.transform(X_df_test)
df_test_good = pd.DataFrame(X_test_pca_good, columns=["PC-" + str(i+1) for i in range(15)])
fp_pca_good_test = "../data/sba_loans_prepared/sba_loans_pca_enc_good_test.csv"
df_test_good.to_csv(fp_pca_good_test, index=False)
```

```python
fp_val = "../data/sba_loans_prepared/sba_loans_num_enc_val.csv"
df_val = pd.read_csv(fp_val)
X_df_val  = df_val[preds]
X_val_pca_good = pca.transform(X_df_val)
df_val_good = pd.DataFrame(X_val_pca_good, columns=["PC-" + str(i+1) for i in range(15)])
fp_pca_good_val = "../data/sba_loans_prepared/sba_loans_pca_enc_good_val.csv"
df_val_good.to_csv(fp_pca_good_val, index=False)
```

```python
X_reduced = pca.fit_transform(X_df_bad)
df_bad_loans_res = pd.DataFrame(pca.explained_variance_ratio_, columns =["exp_var"])
df_bad_loans_res["cum_var_exp"] = df_bad_loans_res["exp_var"].cumsum()
```

```python
fp = "../data/sba_loans_prepared/sba_train_pca_bad_loans.csv"
df_bad_loans_pca = pd.DataFrame(X_reduced)
df_bad_loans_pca.columns = ["PC-" + str(i+1) for i in range(15)]
df_bad_loans_pca.to_csv(fp, index=False)
```

```python
X_test_pca_chgoff = pca.transform(X_df_test)
df_test_chgoff = pd.DataFrame(X_test_pca_chgoff, columns=["PC-" + str(i+1) for i in range(15)])
fp_pca_chgoff_test = "../data/sba_loans_prepared/sba_loans_pca_enc_chgoff_test.csv"
df_test_chgoff.to_csv(fp_pca_chgoff_test, index=False)
```

```python
X_val_pca_chgoff = pca.transform(X_df_val)
df_val_chgoff = pd.DataFrame(X_val_pca_chgoff, columns=["PC-" + str(i+1) for i in range(15)])
fp_pca_chgoff_val = "../data/sba_loans_prepared/sba_loans_pca_enc_chgoff_val.csv"
df_val_chgoff.to_csv(fp_pca_chgoff_val, index=False)
```
